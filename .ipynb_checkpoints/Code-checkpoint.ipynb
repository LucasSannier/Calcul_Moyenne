```python

# info UE1
Liste_UE1 = ['R101','R102','R103','R104','R106','R108','R110','R111','R112','R113','R114','SAE11','SAE12','SAE16']
Notes_UE1 = [12, 14, 15, 10, 11, 13, 14, 12, 15, 9, 11, 14, 16, 13]  # Remplacer par tes notes dans l'ordre
Coefs_UE1 = [10, 10, 7, 7, 5, 6, 5, 4, 2, 5, 5, 20, 20, 7]   # Remplacer par tes coefficients dans l'ordre

# info UE2
Liste_UE2 = ['R101','R103','R104','R105','R110','R111','R112','R113','R114','R115','SAE13','SAE16']
Notes_UE2 = [11, 13, 12, 14, 10, 15, 12, 11, 13, 14, 20, 15]        # Remplacer par tes notes dans l'ordre
Coefs_UE2 = [4, 2, 8, 6, 5, 5, 2, 9, 9, 3, 29, 7]           # Remplacer par tes coefficients dans l'orde

# info UE3
Liste_UE3 = ['R101','R103','R106','R107','R108','R109','R110','R111','R112','R115','SAE14','SAE15','SAE16']
Notes_UE3 = [14, 12, 11, 13, 10, 15, 12, 14, 11, 12, 13, 20, 16]        # Remplacer par tes notes dans l'ordre
Coefs_UE3 = [4, 2, 5, 15, 6, 4, 5, 5, 2, 3, 20, 20, 7]           # Remplacer par tes coefficients dans l'ordre


Numerateur_UE1 = 0
Denominateur_UE1 = 0

for i in range(len(Liste_UE1)):
    note_UE1 = Notes_UE1[i]
    coef_UE1 = Coefs_UE1[i]
    Numerateur_UE1 += note_UE1 * coef_UE1
    Denominateur_UE1 += coef_UE1

if Denominateur_UE1 <= 0 or Numerateur_UE1 < 0 :
    print(f"Vos valeurs sont erronées, il est impossible de calculer {Numerateur_UE1}/{Denominateur_UE1} pour une moyenne comprise entre 0 et 20")
else :
    Moyenne_UE1 = Numerateur_UE1 / Denominateur_UE1
    if Moyenne_UE1 > 20 :
        print("Vous mentez.")
    else:
        print(f"Votre moyenne pour l'UE1 est : {Moyenne_UE1:.2f}")


Numerateur_UE2 = 0
Denominateur_UE2 = 0

for i in range(len(Liste_UE2)):
    note_UE2 = Notes_UE2[i]
    coef_UE2 = Coefs_UE2[i]
    Numerateur_UE2 += note_UE2 * coef_UE2
    Denominateur_UE2 += coef_UE2

if Denominateur_UE2 <= 0 or Numerateur_UE2 < 0 :
    print(f"Vos valeurs sont erronées, il est impossible de calculer {Numerateur_UE2}/{Denominateur_UE2} pour une moyenne comprise entre 0 et 20")
else :
    Moyenne_UE2 = Numerateur_UE2 / Denominateur_UE2
    if Moyenne_UE2 > 20 :
        print("Vous mentez.")
    else:
        print(f"Votre moyenne pour l'UE2 est : {Moyenne_UE2:.2f}")

Numerateur_UE3 = 0
Denominateur_UE3 = 0

for i in range(len(Liste_UE3)):
    note_UE3 = Notes_UE3[i]
    coef_UE3 = Coefs_UE3[i]
    Numerateur_UE3 += note_UE3 * coef_UE3
    Denominateur_UE3 += coef_UE3

if Denominateur_UE3 <= 0 or Numerateur_UE3 < 0 :
    print(f"Vos valeurs sont erronées, il est impossible de calculer {Numerateur_UE3}/{Denominateur_UE3} pour une moyenne comprise entre 0 et 20")
else :
    Moyenne_UE3 = Numerateur_UE3 / Denominateur_UE3
    if Moyenne_UE3 > 20 :
        print("Vous mentez.")
    else:
        print(f"Votre moyenne pour l'UE3 est : {Moyenne_UE3:.2f}")
```
