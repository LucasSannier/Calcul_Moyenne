```python

Numerateur_UE1 = 0
Denominateur_UE1 = 0

Numerateur_UE2 = 0
Denominateur_UE2 = 0

Numerateur_UE3 = 0
Denominateur_UE3 = 0

Liste_UE1 =['R101','R102','R103','R104','R106','R108','R110','R111','R112','R113','R114','SAE11','SAE12','SAE16',]
for i,elt in enumerate(Liste_UE1):
    note_UE1 = int(input("Pour le module de lUE1 {}, quelle est votre note ?".format(elt)))
    coef_UE1 = int(input("Pour le module de lUE1 {}, quelle est le coéfficient ?".format(elt)))
    Numerateur_UE1 += note_UE1*coef_UE1
    Denominateur_UE1 += coef_UE1
    
Liste_UE2 =['R101','R103','R104','R105','R110','R111','R112','R113','R114','R115','SAE13','SAE16']
for i,elt in enumerate(Liste_UE2):
    note_UE2 = int(input("Pour le module de lUE2 {}, quelle est votre note ?".format(elt)))
    coef_UE2 = int(input("Pour le module de lUE2 {}, quelle est le coéfficient ?".format(elt)))
    Numerateur_UE2 += note_UE2*coef_UE2
    Denominateur_UE2 += coef_UE2
    
Liste_UE3 =['R101','R103','R106','R107','R108','R109','R110','R111','R112','R115','SAE14','SAE15','SAE16']
for i,elt in enumerate(Liste_UE3):
    note_UE3 = int(input("Pour le module de lUE3 {}, quelle est votre note ?".format(elt)))
    coef_UE3 = int(input("Pour le module de lUE3 {}, quelle est le coéfficient ?".format(elt)))
    Numerateur_UE3 += note_UE3*coef_UE3
    Denominateur_UE3 += coef_UE3

if Denominateur_UE1 <= 0 or Numerateur_UE1 < 0 :
    print(f"Vos valeurs sont érroné, il est impossible de calculer {Numerateur_UE1}/{Denominateur_UE1} pour une moyenne comprise entre 0 et 20")
else :
    Moyenne_UE1 = Numerateur_UE1/Denominateur_UE1
    if Moyenne_UE1 > 20 :
        print("Vous mentez.")
    else:
        print(f"Votre moyenne pour l'UE1 est : {Moyenne_UE1:.2f}")


if Denominateur_UE2 <= 0 or Numerateur_UE2 < 0 :
    print(f"Vos valeurs sont érroné, il est impossible de calculer {Numerateur_UE2}/{Denominateur_UE2} pour une moyenne comprise entre 0 et 20")
else :
    Moyenne_UE2 = Numerateur_UE2/Denominateur_UE2
    if Moyenne_UE2 > 20 :
        print("Vous mentez.")
    else:
        print(f"Votre moyenne pour l'UE2 est : {Moyenne_UE2:.2f}")


if Denominateur_UE3 <= 0 or Numerateur_UE3 < 0 :
    print(f"Vos valeurs sont érroné, il est impossible de calculer {Numerateur_UE3}/{Denominateur_UE3} pour une moyenne comprise entre 0 et 20")
else :
    Moyenne_UE3 = Numerateur_UE3/Denominateur_UE3
    if Moyenne_UE3 > 20 :
        print("Vous mentez.")
    else:
        print(f"Votre moyenne pour l'UE3 est : {Moyenne_UE3:.2f}")
```
