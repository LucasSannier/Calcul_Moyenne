```python
import tkinter as tk
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

liste = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

Coefs_UE1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
Coefs_UE2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
Coefs_UE3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]
saisie = {}

def affich():
    no = [float(saisie[i].get()) for i in liste]
    m_ue1 = np.average(no, weights=Coefs_UE1)
    m_ue2 = np.average(no, weights=Coefs_UE2)
    m_ue3 = np.average(no, weights=Coefs_UE3)
    resultat = np.mean([m_ue1, m_ue2, m_ue3])
    texte2.config(text=f"Résultat : {resultat:.2f}")
    
    UE = ["UE1", "UE2", "UE3"]
    Moyenne = [m_ue1, m_ue2, m_ue3]
    ax.clear() 
    couleurs = ["red" if m < 8 else "green" if m >= 10 else "orange" for m in Moyenne]
    ax.bar(UE, Moyenne, color = couleurs)
    ax.set_title("Voici votre moyenne pour chaque UE")
    ax.set_xlabel("UE")
    ax.set_ylabel("Moyenne")
    ax.set_ylim(0, 20)

    canvas_matplotlib.draw()
    
    if any(m < 8 for m in Moyenne):
        print("Votre année est invalidée car l'une de vos UE est inférieure à 8.")
    elif all(m >= 10 for m in Moyenne):
        print("Vous passez en deuxième année !")
    else:
        print("Vous passez en deuxième année mais vous devrez valider vos UE < 10 aux rattrapages.")
        
    return resultat

fenetre = tk.Tk()
fenetre.geometry("1000x500")
fenetre.title("Semestre 1")

texte1 = tk.Label(fenetre, text="Entrez vos notes dans les modules correspondants pour calculer votre moyenne")
texte1.pack(pady=10)
texte2 = tk.Label(fenetre, text="")
texte2.pack(pady=5)

#===============================================================SCROLLBAR=(IA)===========================================================
canvas = tk.Canvas(fenetre)
scrollbar = tk.Scrollbar(fenetre, orient="vertical", command=canvas.yview)
scrollable_frame = tk.Frame(canvas)

scrollable_frame.bind(
    "<Configure>",
    lambda e: canvas.configure(scrollregion=canvas.bbox("all"))
)

canvas.create_window((0, 0), window=scrollable_frame, anchor="nw")
canvas.configure(yscrollcommand=scrollbar.set)

canvas.pack(side="left", fill="both", expand=True)
scrollbar.pack(side="right", fill="y")
#=======================================================================================================================================

for i in liste:
    ligne = tk.Frame(scrollable_frame)
    ligne.pack(pady=2)
    
    texte = tk.Label(ligne, text=i, width=8, anchor="e")
    texte.pack(side=tk.LEFT, padx=5)
    
    saisie[i] = tk.Entry(ligne, width=10)
    saisie[i].pack(pady=2)

#===============================================================TABLEAU=TK=(IA)=========================================================
fig, ax = plt.subplots(figsize=(5, 3.5))
canvas_matplotlib = FigureCanvasTkAgg(fig, master=fenetre)
canvas_matplotlib.get_tk_widget().pack(side="right", fill="both", expand=True, padx=20)
#=======================================================================================================================================

btn = tk.Button(fenetre, text="Voir votre Moyenne pour chaque UE", command=affich)
btn.pack(pady=5)

fenetre.mainloop()
```

```python
## import matplotlib.pyplot as plt
import numpy as np
import matplotlib.pyplot as plt

liste = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

Coefs_UE1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
Coefs_UE2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
Coefs_UE3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

coefficients = {"UE1": dict(zip(liste, Coefs_UE1)),"UE2": dict(zip(liste, Coefs_UE2)),"UE3": dict(zip(liste, Coefs_UE3))}

notes = {}

for i in liste:
    notes[i] = float(input(f"quelle est votre note pour {i} ? "))

def calculer_moyenne(notes, coefs):
    no = [notes[i] for i in liste]
    co = [coefs[i] for i in liste]
    return np.average(no, weights=co)

Moyenne_UE1 = calculer_moyenne(notes, coefficients["UE1"])
Moyenne_UE2 = calculer_moyenne(notes, coefficients["UE2"])
Moyenne_UE3 = calculer_moyenne(notes, coefficients["UE3"])

if not (0 <= Moyenne_UE1 <= 20 and 0 <= Moyenne_UE2 <= 20 and 0 <= Moyenne_UE3 <= 20):
    print("Vos valeurs sont erronées elles doivent toutes être comprises entre 0 et 20.")
else:
    UE = ["UE1", "UE2", "UE3"]
    Moyenne = [Moyenne_UE1, Moyenne_UE2, Moyenne_UE3]
    couleurs = ["red" if m < 8 else "green" if m >= 10 else "orange" for m in Moyenne]
    plt.bar(UE, Moyenne, color = couleurs)
    plt.title("Voici votre moyenne pour chaque UE")
    plt.xlabel("UE")
    plt.ylabel("Moyenne")
    plt.show()
    if any(m < 8 for m in Moyenne):
        print("Votre année est invalidée car l'une de vos UE est inférieure à 8.")
    elif all(m >= 10 for m in Moyenne):
        print("Vous passez en deuxième année !")
    else:
        print("Vous passez en deuxième année mais vous devrez valider vos UE < 10 aux rattrapages.")
```
