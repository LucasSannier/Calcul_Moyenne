# CALCUL MOYENNE BUT 1 

```python
import numpy as np

liste = [f"R{i}" for i in range (101,116)] + [f"SAE{i}" for i in range (11, 17)]
Coefs_UE1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
Coefs_UE2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
Coefs_UE3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]
Notes = [float(input(f"quelle est votre note pour {elt} ?")) for elt in (liste)]

Notes_np = np.array(Notes)

Moyenne_UE1 = np.average(Notes_np, weights=Coefs_UE1)
Moyenne_UE2 = np.average(Notes_np, weights=Coefs_UE2)
Moyenne_UE3 = np.average(Notes_np, weights=Coefs_UE3)


if not (0 <= Moyenne_UE1 <= 20 and 0 <= Moyenne_UE2 <= 20 and 0 <= Moyenne_UE3 <= 20):
    print("Vos valeurs sont erronées, elles doivent toutes être comprises entre 0 et 20.")
else :
    print(f"Votre moyenne pour l'UE1 est : {Moyenne_UE1:.2f}")
    print(f"Votre moyenne pour l'UE1 est : {Moyenne_UE2:.2f}")
    print(f"Votre moyenne pour l'UE1 est : {Moyenne_UE3:.2f}")
```
